# Lesson 21 - Pick It Up (Claw Basics)

Some robots have an extra part: a **robotic arm with a claw** on the end (the
"ultimate kit"). In this lesson you learn to control it with `myRobot.arm`.

**Not every robot has a claw.** On a robot with no arm, the arm commands are
still safe — they just do nothing. So you can run this whole lesson on any
robot; you only see the arm move on a robot that actually has one.

## 1. Start The Robot

Run this first.

```python
from lesson_header import *
import time

show_v2_status()
myRobot = bot(base_speed=300)
```

## 2. Does this robot have a claw? Test it first.

Run the self-test and **watch the arm**. The claw should open, close, and open
again, then the arm should raise, lower, and settle in the middle. It prints
each step.

```python
myRobot.arm.test()
```

```python
# You can also just ask whether an arm is present:
if myRobot.arm.available:
    print("This robot has a claw!")
else:
    print("No claw on this robot - arm commands will do nothing.")
```

## 3. Open and close the claw

Try them one at a time so you can see what each does.

```python
myRobot.arm.open_gripper()    # claw opens
time.sleep(1)
myRobot.arm.close_gripper()   # claw closes (grips)
```

## 4. Raise and lower the arm

```python
myRobot.arm.lift_up()
time.sleep(1)
myRobot.arm.lift_down()
time.sleep(1)
myRobot.arm.ready()           # tidy pose: gripper open, arm at middle height
```

## 5. Grab and place

Put a **light, soft** object (a foam ball or small block) just in front of the
open claw. `grab()` does the whole pick-up for you: **open -> lower -> close ->
raise.**

```python
myRobot.arm.grab()
```

`place()` puts it back down: **lower -> open -> raise.**

```python
myRobot.arm.place()
```

## 6. Challenge: carry something across the room

Grab an object, drive somewhere, then put it down in a new spot.

```python
myRobot.arm.grab()
myRobot.move.forward(seconds=1.0, speed=80)
myRobot.arm.place()
```

## Tips
- If the claw closes too early, slow the sequence down:
  `myRobot.arm.grab(settle=1.0)` (a longer pause between each step).
- Only grab light, soft objects — the claw is not strong.
- If "lift up" and "lift down" look swapped on your robot, tell your teacher —
  it is a one-line fix in the arm library.
